```python
distance_url = "https://raw.githubusercontent.com/Trickwillfrit/sustainable-logistics/main/Distance_matrix.csv"
population_url = "https://raw.githubusercontent.com/Trickwillfrit/sustainable-logistics/main/Pop_district.xlsx"
```

```python
import pandas as pd
import numpy as np

distance_df = pd.read_csv(distance_url, encoding='latin1')
population_df = pd.read_excel(population_url, sheet_name="Total")

print(distance_df.shape)
print(population_df.shape)

display(distance_df.head())
display(population_df.head())
```

```python
# Basic information about the two datasets

print("Distance matrix shape:", distance_df.shape)
print("Population data shape:", population_df.shape)

print("\nDistance matrix columns:")
print(distance_df.columns.tolist())

print("\nPopulation data columns:")
print(population_df.columns.tolist())

print("\nFirst rows of distance matrix:")
display(distance_df.head())

print("\nFirst rows of population data:")
display(population_df.head())
```

```python
# --- Clean distance matrix ---

# Drop useless index column if it exists
distance_clean = distance_df.drop(columns=["I"], errors="ignore").copy()

# Rename commune column
distance_clean = distance_clean.rename(columns={"ID": "commune"})

# Remove spaces before/after commune names
distance_clean["commune"] = distance_clean["commune"].astype(str).str.strip()

# Fix one remaining encoding issue in hospital name
distance_clean = distance_clean.rename(
    columns={"Hôpital de la VallÃ©e de Joux": "Hôpital de la Vallée de Joux"}
)

# Identify hospital columns
hospital_cols = [col for col in distance_clean.columns if col != "commune"]


# --- Clean population data ---

population_clean = population_df.copy()

# Keep only real commune rows: rows where the commune code exists
population_clean = population_clean[population_clean["Unnamed: 2"].notna()].copy()

# Keep useful columns only
population_clean = population_clean[
    ["District ", "Mesures", "Unnamed: 2", "Population au 31.12..8"]
].copy()

# Rename columns
population_clean = population_clean.rename(
    columns={
        "District ": "district",
        "Mesures": "commune",
        "Unnamed: 2": "commune_code",
        "Population au 31.12..8": "population_2025"
    }
)

# Clean commune names
population_clean["commune"] = population_clean["commune"].astype(str).str.strip()

# Make population numeric
population_clean["population_2025"] = pd.to_numeric(
    population_clean["population_2025"], errors="coerce"
)


# --- Print results ---

print("Distance clean shape:", distance_clean.shape)
print("Population clean shape:", population_clean.shape)

print("\nNumber of communes in distance matrix:", distance_clean["commune"].nunique())
print("Number of communes in population file:", population_clean["commune"].nunique())

print("\nNumber of hospitals:", len(hospital_cols))

print("\nDistance clean preview:")
display(distance_clean.head())

print("\nPopulation clean preview:")
display(population_clean.head())
```

# Data validation: checking commune names

Before building the optimization model, we need to verify that the commune names are consistent between the two datasets:

1. The distance matrix, which contains the distance from each commune to each hospital.
2. The population dataset, which contains the district and population of each commune.

This step is important because later we will merge both datasets using the commune name.  
If some names do not match, some communes could disappear from the final dataset, and the optimization model would be incorrect.

In this cell, we check:

- whether there are duplicated commune names in the population dataset;
- whether some communes exist in the population dataset but not in the distance matrix;
- whether some communes exist in the distance matrix but not in the population dataset.

```python
# DATA VALIDATION: CHECK COMMUNE NAMES


# Check duplicated commune names in the population data
duplicated_communes = population_clean[
    population_clean["commune"].duplicated(keep=False)
].sort_values("commune")

print("Duplicated commune rows in population data:")
display(duplicated_communes)


# Check communes that exist in population data but not in distance matrix
communes_population = set(population_clean["commune"])
communes_distance = set(distance_clean["commune"])

missing_in_distance = sorted(communes_population - communes_distance)

print("\nCommunes in population file but missing in distance matrix:")
print(missing_in_distance)
print("Number:", len(missing_in_distance))


# Check communes that exist in distance matrix but not in population data
missing_in_population = sorted(communes_distance - communes_population)

print("\nCommunes in distance matrix but missing in population file:")
print(missing_in_population)
print("Number:", len(missing_in_population))
```

# Merge population data with distance matrix

Now that the commune names are aligned, we can merge the two datasets into one final dataset.

The goal is to create one table where each row represents one commune and contains:

- the district of the commune;
- the commune name;
- the commune population in 2025;
- the distance from this commune to each hospital.

This merged dataset will be the basis for the exploratory analysis and for the optimization model.

```python
# MERGE POPULATION DATA WITH DISTANCE MATRIX

# Merge the population data with the distance matrix using the commune name
data = population_clean.merge(
    distance_clean,
    on="commune",
    how="inner"
)

# Check the shape of the merged dataset
print("Merged dataset shape:", data.shape)

# Check that no commune was lost during the merge
print("Number of communes in merged dataset:", data["commune"].nunique())

# Check total population
print("Total population in merged dataset:", data["population_2025"].sum())

# Display the first rows
display(data.head())
```

# First EDA: closest hospital for each commune

In this cell, we identify the closest hospital for each commune.

For every commune, the distance matrix gives the distance to each hospital.  
We use this information to find:

- the closest hospital;
- the distance to the closest hospital;
- the distance converted from meters to kilometers.

This helps us understand the current accessibility of hospitals in the canton of Vaud before running the optimization model.

```python
# FIRST EDA: CLOSEST HOSPITAL FOR EACH COMMUNE

# Hospital columns are all columns after the population information
# We exclude district, commune, commune_code, and population_2025
hospital_cols = [
    col for col in data.columns
    if col not in ["district", "commune", "commune_code", "population_2025"]
]

# For each commune, find the hospital with the minimum distance
data["closest_hospital"] = data[hospital_cols].idxmin(axis=1)

# For each commune, find the distance to the closest hospital
data["closest_distance_m"] = data[hospital_cols].min(axis=1)

# Convert distance from meters to kilometers
data["closest_distance_km"] = data["closest_distance_m"] / 1000

# Display useful columns
closest_hospital_eda = data[
    ["district", "commune", "population_2025", "closest_hospital", "closest_distance_km"]
].copy()

display(closest_hospital_eda.head(20))

# Summary statistics
print("Average closest distance (km):", round(data["closest_distance_km"].mean(), 2))
print("Median closest distance (km):", round(data["closest_distance_km"].median(), 2))
print("Maximum closest distance (km):", round(data["closest_distance_km"].max(), 2))
```

# EDA: closest hospitals by district

In this cell, we analyze hospital accessibility at the district level.

For each commune, we already identified the closest hospital.  
Now we group the results by district and by closest hospital.

This allows us to see, for each district:

- which hospitals are the closest to the communes of that district;
- how many communes are naturally assigned to each hospital;
- how much population is served by each hospital;
- the average and maximum distance to the closest hospital.

This step is useful before optimization because our project aims to evaluate whether there are too many hospitals in some districts and which hospitals are the most important to keep.

```python
# EDA: CLOSEST HOSPITALS BY DISTRICT

# Group communes by district and closest hospital
district_hospital_eda = (
    data
    .groupby(["district", "closest_hospital"])
    .agg(
        number_of_communes=("commune", "count"),
        population_served=("population_2025", "sum"),
        average_distance_km=("closest_distance_km", "mean"),
        max_distance_km=("closest_distance_km", "max")
    )
    .reset_index()
)

# Sort results by district and then by population served
district_hospital_eda = district_hospital_eda.sort_values(
    by=["district", "population_served"],
    ascending=[True, False]
)

# Display the district-level accessibility table
display(district_hospital_eda)

# Summary: number of closest hospitals per district
hospitals_per_district_eda = (
    district_hospital_eda
    .groupby("district")
    .agg(
        number_of_closest_hospitals=("closest_hospital", "count"),
        total_population=("population_served", "sum"),
        total_communes=("number_of_communes", "sum")
    )
    .reset_index()
    .sort_values("number_of_closest_hospitals", ascending=False)
)

print("Summary by district:")
display(hospitals_per_district_eda)
```

# District-level accessibility summary

The previous table showed hospital accessibility by district and by closest hospital.

Now we summarize the results at the district level only.  
This helps us compare districts and identify which areas are currently more or less accessible.

For each district, we calculate:

- the number of communes;
- the total population;
- the average distance to the closest hospital;
- the maximum distance to the closest hospital;
- the population-weighted average distance.

The population-weighted average distance is important because a commune with many inhabitants should have more influence than a commune with few inhabitants.

```python
# CELL 7 — DISTRICT-LEVEL ACCESSIBILITY SUMMARY

# Calculate weighted distance for each commune
# This gives more importance to communes with larger populations
data["weighted_closest_distance"] = (
    data["closest_distance_km"] * data["population_2025"]
)

# Create district-level summary
district_accessibility = (
    data
    .groupby("district")
    .agg(
        number_of_communes=("commune", "count"),
        total_population=("population_2025", "sum"),
        average_closest_distance_km=("closest_distance_km", "mean"),
        max_closest_distance_km=("closest_distance_km", "max"),
        total_weighted_distance=("weighted_closest_distance", "sum")
    )
    .reset_index()
)

# Population-weighted average distance
district_accessibility["population_weighted_avg_distance_km"] = (
    district_accessibility["total_weighted_distance"]
    / district_accessibility["total_population"]
)

# Drop technical column
district_accessibility = district_accessibility.drop(
    columns=["total_weighted_distance"]
)

# Sort by weighted average distance, from least accessible to most accessible
district_accessibility = district_accessibility.sort_values(
    by="population_weighted_avg_distance_km",
    ascending=False
)

display(district_accessibility)

print(
    "Global population-weighted average closest distance (km):",
    round(
        (data["closest_distance_km"] * data["population_2025"]).sum()
        / data["population_2025"].sum(),
        2
    )
)
```
